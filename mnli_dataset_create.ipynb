# MNLI Test



```python
import pandas as pd

Ag_news_test = pd.read_csv("Data_Raw_File/raw/ag_news/ag_news_csv/test.csv",sep = ",",header=None)
Ag_news_test.columns = ["Article Class", "Title", "Description"]


mapping_mask = {1:"World" ,2:"Sports",3:"Business",4:"Sci/Tech"}
Ag_news_test["Article Class"] =  Ag_news_test["Article Class"].map(mapping_mask)


Ag_news_test



```

```python
import numpy as np

# text id indicate premise  {0:"World" ,1:"Sports",2:"Business",3:"Sci/Tech"}


premise = np.repeat(Ag_news_test["Description"],4)

hypothesis = ["This example is World","This example is Sports","This example is Bussiness","This example is Sci/Tech"] * len(Ag_news_test)

# https://huggingface.co/docs/transformers/model_doc/roberta
# roberta format
# text = [f"<s> {premise} </s><s> {hypothesis} </s>" for premise, hypothesis in zip(premise, hypothesis)]

# https://huggingface.co/docs/transformers/en/model_doc/deberta

# text = [f"[CLS]  {premise} [SEP] {hypothesis} [SEP]" for premise, hypothesis in zip(premise, hypothesis)]

# Sanity Check
print(len(premise))
print(len(hypothesis))
# print(len(text))

data = { "textid" : [0,1,2,3] * len(Ag_news_test),
        "class" : ["World" ,"Sports","Business","Sci/Tech"] * len(Ag_news_test),
        "text":  premise,
        "pair": hypothesis,
        "target" : list(np.repeat(range(0,(len(Ag_news_test))),4))
}


df_text= pd.DataFrame(data).head(20)

# sanity check for the format of text
print(df_text["text"].iloc[1]) 
df_text.to_csv("NLP_dataset/MNLI_formatted/Shared_Test_MNLI.tsv",sep="\t",index=False)
```
